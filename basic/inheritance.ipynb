# Inheritance


```python
class Vechial():
    def __init__(self , make  , model):
        self.make = make
        self.model = model


       #getter function of make
        
        @property
        def make(self):
            return self.make

        #setter function for make

        @make.setter
        def make(self , new_make):
            self.make = new_make
              
        #getter function of make 

        @property
        def model(self):
            return self.model  

        #setter function for make

        @model.setter
        def model(self, new_model):
            self.model = new_model     
        
        
    def print_info(self):
        print(f"{self.make} {self.model}")
   
             
```

```python
suzuki = Vechial('maruthi' , 'swift')
suzuki.print_info()
```

```python
suzuki.make = 'accord'
suzuki.print_info()
```

```python
class Truck(Vechial):
    def print_info(self):
        print("this is a truck")
        return super().print_info()

class Bike(Vechial):
    def print_info(self):
        print("this is a car")
        return super().print_info()

class Car(Vechial):
    def print_info(self):
        print("this is a car")
        return super().print_info()                
```

```python

eicher = Truck('TATA' , 'eicher')
eicher.print_info()
tata = Car('altroz' , '5star')
tata.print_info()
```

```python
tata = Truck('Tata', 'eicher')
tata.print_info()
enfield = Bike('Eicher', 'Royal Enfield')
enfield.print_info()
city = Car('Honda', 'city')
city.print_info()
```

```python
class Animal():
    def __init__(self, carnivours,heribvours,Terrestrial,aquatic):
        self.carnivours = carnivours
        self.heribvours = heribvours
        self.Terrestrial = Terrestrial
        self.aquatic = aquatic

    def print_info(self):
        print(f"{self.carnivours} {self.aquatic} {self.heribvours} {self.Terrestrial}")    
```

```python
flesh = Animal('lion','tiger','dog','shark')
flesh.print_info()
```

```python
flesh.carnivours = 'pig'
flesh.print_info()
```
