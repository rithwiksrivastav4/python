```python
numbers = ['4','5','9','6','2']

def sort_by_length(numbers):
    # Use sorted() instead of sort() since sort() is a method for list objects
    list1 = sorted(numbers, key=len, reverse=True)
    return list1

# Call the function and print the result
sorted_numbers = sort_by_length(numbers)
print(sorted_numbers)
```

```python
colours = ['red', 'green', 'yellow', 'pink','a', 'black', 'gold']

def sort_by_length_desc(colours):
    # Sort the colours by length in descending order
    list1 = sorted(colours, key=len, reverse=False)
    return list1

# Call the function and print the result
sorted_colours_desc = sort_by_length_desc(colours)
print(sorted_colours_desc)
```
