```python
NUMBER = 13195
for factor in range(NUMBER-1,1,-1):
    # from 13194 to 2 find factors
    if NUMBER % factor == 0:
        # check if this factor is prime or not
        is_prime = True
        for index in range(2,factor):
            if factor % index == 0:
               is_prime = False
               break
        # if it is prime print and exit
        if is_prime:
                 print(factor)
                 break
```

```python
def revers(text):
    cleaned_text = "".join(char.lower() for char in text if char.isalnum())
    if cleaned_text == cleaned_text[::-1]:
        print("is polindrome")
    else:
        print("is not polindrome") 
    return cleaned_text == cleaned_text[::-1]      
```

```python
# here "x" is the parameter ,parameter: The value while we defining the function
def even( x ):
    '''finding even and odd numbers from the list and adding both even and odd numbers to the new list
    '''
   # list = [2,3,4,5,6,7,8,23,45,44,33,22]
    eve = []
    odd = []
    # iterating form the list
    for i in x:
      # if the given condition is true then add to the even list
      if i % 2 == 0:
          eve.append(i)
          #print(f"{i}:is even")
      else:
          # if the condition is not true then add it to odd list
          odd.append(i)
          #print(f"{i}:is odd") 
    return eve,odd      
#even(list)           
```

```python
def unique(u):
    '''finding unique and prime numbers from the list and adding it to a new list,
    and finding non unique number and not prime number from the list and adding it to a new list '''
    uni = []
    notuni = []
    for i in u:
        if i not in uni and i % 2 ==0 :
            uni.append(i)
        elif i not in notuni and i % 2 !=0:
            notuni.append(i)    
    return uni,notuni        
```
