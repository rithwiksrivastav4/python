```python
import random
import string

def start_game():
    min_char = 'A'
    max_char = 'Z'
    max_guess_count = 3

    # Random letter from A to Z
    random_letter = random.choice(string.ascii_uppercase)
    guessed = False

    print(f"🔤 Guess the letter I'm thinking of (between {min_char} and {max_char})!")
    
    for guess_count in range(max_guess_count):
        user_input = input(f"Attempt {guess_count + 1}: Enter a letter: ").upper()

        # Validate input
        if len(user_input) != 1 or user_input not in string.ascii_uppercase:
            print("❌ Invalid input! Please enter a single letter (A–Z).")
            continue

        # Check guess
        if user_input == random_letter:
            print("🎉 Wonderful! You guessed it right!")
            guessed = True
            break
        elif user_input < random_letter:
            print("🔼 You're close! Try a letter later in the alphabet.")
        else:
            print("🔽 You're close! Try a letter earlier in the alphabet.")

    # After all guesses
    if not guessed:
        print(f"😢 You lost. The correct letter was '{random_letter}'. Better luck next time!")

if __name__ == '__main__':
    start_game()
```
