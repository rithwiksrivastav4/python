```python
class Vehicle:
    quantity = 0
    cars = set()
    def __init__(self, make, model):
        self.__make = make
        self.__model = model

        
        #if not in quantity them add 
        #Vehicle.cars.append(self.make)
        #Vehicle.mod.append(self.model)
        if make not in Vehicle.cars:
            Vehicle.cars.add(make)
            Vehicle.quantity += 1

    # Getter 
    @property
    def make(self):
        return self.__make
    # setter
    @make.setter
    def make(self, new_make):
        if not isinstance(new_make, str):
            raise ValueError("Model must be string:")
        self.__make = new_make

        if new_make != self.__make:
            Vehicle.cars.discard(self.__make)
            Vehicle.cars.add(new_make)
            self.__make = new_make
    
    @property
    def model(self):
        return self.__model
    
    @model.setter
    def model(self, new_model):
        self.__model = new_model

        if new_model != self.__model:
            Vehicle.cars.discard(self.__model)
            Vehicle.cars.add(new_model)
            self.__model = new_model
    
    def print_info(self):
        print(f"{self.__make} {self.__model}")
```

```python
cityy = Vehicle('Honda', 'city')
cityy.model = 'Accord' # sets values calls setter automatically
mar = Vehicle("maruti","suzuki")
print(cityy.make) # calling gatter automatically
#cityy.model() # if we use @property  we can't use like this citty.make(), it only works for normal methods
#cityy.make = "123"
mar.print_info()
cityy.print_info()

print(Vehicle.cars)
print(mar.model)
Vehicle.quantity
```

```python
class Human():

    population = 0
    data = set()

    def __init__(self,name,age, alive = True):

        self.name = name
        self.age = age
        self.alive = alive
        
        #Human.population += 1
       # Human.data.append(self.name)

        if self.name not in Human.data:
            Human.data.add(name)
            Human.population += 1

    def greet(self):
        print(f"hey my name is :{self.name}. good morning,iam  :{self.age} old")

    def die(self):
        if self.alive:
            print(self.name ,":this person is no more")
            Human.population -= 1
            self.alive = False
            Human.data.discard(self.name)
        else:
            print(self.name ,":this person is already dead") 

    def child(self , number):
        Human.population += number

class Employee(Human):

    employees = 0
    names = set()
    """this is child class and taking attributes from parent class"""
    def __init__(self,name,age, company, post,employed = True):
        super().__init__( name, age )
        self.company = company
        self.post = post
        self.employed = employed

        if self.name and self.employed not in Employee.names:
            Employee.names.add(name)
            Employee.employees += 1
    def work(self):
        print(self.name, "is working in:",self.company ,"as a :",self.post,"and his age is ", self.age)

    def leave_company(self):
        if self.employed: 
            print(self.name,"is left from the company")
            Employee.employees -= 1
            Employee.names.discard(self.name)
            self.employed = False
        else:
            print(self.name, "is already left from the company")    

```
