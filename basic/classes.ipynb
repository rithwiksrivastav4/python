# getter and setter

```python
class Vechial():
    def __init__(self , make  , model):
        self.__make = make
        self.__model = model


       #getter function of make
        
        @property
        def make(self):
            return self.__make

        #setter function for make

     
        def make(self , new_make):
            self.__make = new_make
              
        #getter function of make 

        @property
        def model(self):
            return self.__model  

        #setter function for make

       
        def model(self, new_model):
            self.__model = new_model     
        
        
    def print_info(self):
        print(f"{self.__make} {self.__model}")
   
             
```

```python
suzuki = Vechial('maruthi' , 'swift')
tata = Vechial('altroz' , '5star')
suzuki.__make
```

```python
suzuki.make = 'accord'
suzuki.print_info()
tata.print_info()
```

```python
class Vehicle:
    def __init__(self, make, model):
        self.__make = make
        self.__model = model

    # Getter 
    @property
    def make(self):
        return self.__make
    # setter
    @make.setter
    def make(self, new_make):
        self.__make = new_make
    
    @property
    def model(self):
        return self.__model
    
    @model.setter
    def model(self, new_model):
        self.__model = new_model
    
    def print_info(self):
        print(f"{self.__make} {self.__model}")
```

```python
city = Vehicle('Honda', 'city')
city.model = 'Accord'
city.print_info()
```
